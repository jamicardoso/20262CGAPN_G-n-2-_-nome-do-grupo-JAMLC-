```python

## Parâmetro
valorCreche = 1.57 
valorPreEscola = 0.82 
valorFundamental = 0.57


#Custo dos alimentos por refeição(R$)
#Fonte: https://static.even3.com/anais/1072522.pdf
custo_alimentos_refeicao = 3.60


## Funções (processo do backend)
def ler_numero(mensagem): 
    numero = input(mensagem)
    while not numero.isdigit() : 
        print("Você digitou o número incorreto")
        numero = input(mensagem)
    numero = int(numero)
    return numero 

def classificar_porte(numMatriculas):
    if numMatriculas <=50 : 
        porteEscola = "Micro"
    elif numMatriculas <= 200 : 
        porteEscola = "Pequena"
    elif numMatriculas <= 400: 
        porteEscola = "Média"
    elif numMatriculas <=1000: 
        porteEscola = "Grande"
    else: 
        porteEscola = "Macro"
    return porteEscola

def calcular_repasse(QtdeMatricula,valorUnitario,dias) : 
    TotalRepasse = QtdeMatricula * valorUnitario * dias
    return TotalRepasse
    
def calcular_custo_anual(QtdeMatricula,dias,custo_alimentos_refeicao):
    TotalCusto = QtdeMatricula * dias * custo_alimentos_refeicao
    return TotalCusto

def calcula_cobertura(repasseTotal,custoTotal):
    if custoTotal ==0:
       return 0,  0
    cobertura =(repasseTotal/custoTotal)*100
    if repasseTotal >=custoTotal:
        diferenca = 0 
        print("O repasse cobre o custo estimado dos alimentos")
    else:
        diferenca = custoTotal-repasseTotal
    return cobertura,diferenca

### Programa Principal
print("Olá, seja bem-vindo e bem-vinda ao simulador PNAE")
print()
print("Dados da Escola")
nome = input("Digite seu nome:")
escola = input("Digite o nome da escola: " )

## Pedir dias letivos
diasletivos = ler_numero("Qual a quantidade de dias letivos do ano : ")
print()


### Pedir as Matrículas ###
print("Digitação das Matrículas")
creche = ler_numero("Digite as matrículas na creche : ")
preescola = ler_numero("Digite as matrículas da pré escola: ")
fundamental = ler_numero( "Digite as matrículas do fundamental: ")

print()

totalmatriculas = creche + preescola + fundamental

repasseCreche = calcular_repasse(creche,valorCreche,diasletivos)
repassePreEscola = calcular_repasse(preescola,valorPreEscola,diasletivos)
repasseFundamental = calcular_repasse(fundamental,valorFundamental,diasletivos)

repasseTotal = repasseCreche + repassePreEscola + repasseFundamental

custoTotal= calcular_custo_anual(totalmatriculas,diasletivos,custo_alimentos_refeicao)

#Chamando a função de cobertura 
cobertura,diferenca =calcula_cobertura(repasseTotal,custoTotal)

porte = classificar_porte(totalmatriculas)

##Montando a tela com resultados usando função print
print( '----Cálculos ---------') 
print()
print("Sua escola é de Porte: ", porte)
print("Valor a receber Creche: ", repasseCreche)
print("Valor a receber Pré Escola : ", repassePreEscola)
print("Valor a receber Fundamental : ", repasseFundamental) 
print("Valor a receber Total : " , repasseTotal)
print("Valor do Custo Total : ",custoTotal)
print("Valor da Cobertura Anual(%) : ",cobertura)
print("Valor da Diferença do Custo Total - Repasse Total : ", diferenca) 
print('--- Fim dos Cálculos ------')
print()
input("Digite Enter para seguir ") 
```
